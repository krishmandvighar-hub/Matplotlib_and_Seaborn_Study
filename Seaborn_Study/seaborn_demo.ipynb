```python
# 1. Lineplot
# Lineplot is used to show the relationship between two continuous variables.
# It is commonly used for time-series or ordered data.

# Example: Visualizing trend over time

import seaborn as sns
import matplotlib.pyplot as plt

fmri=sns.load_dataset("fmri")
sns.lineplot(x="timepoint",y="signal",hue="region",data=fmri)
plt.show()

# Explanation: x="timepoint" sets time on x-axis, y="signal" values plotted on y-axis
# and hue="region" separates lines by region
```

```python
# 2. Scatterplot
# Scatterplot is used to show the relationship between two numerical variables.
# It helps in identifying patterns or correlations.

# Example: Relationship between total bill and 

tips=sns.load_dataset("tips")
sns.scatterplot(x="total_bill",y="tip",hue="day",data=tips)
plt.show()

# Explanation: x, y numerical variables and hue="day" adds color grouping
```

```python
# 3. Barplot
# Barplot is used to compare numerical values across categories. 
# It shows the average value with confidence intervals.

# Example: Average total bill per day

sns.barplot(x="day",y="total_bill",data=tips)
plt.show()

# Explanation: x="day" categories and y="total_bill" aggregated values (mean by default)
```

```python
# 4. Boxplot
# Boxplot is used to show the distribution of data based on quartiles.
# It helps identify spread and outliers.
# Example: Distribution of total bill by day

sns.boxplot(x="day",y="total_bill",data=tips)
plt.show()
```

```python
# 5. Histplot
# Histplot is used to visualize the distribution of a single numerical variable.

# Example: Distribution of total bill

sns.histplot(tips["total_bill"],kde=True)
plt.show()

# Explanation: kde=True adds density curve
```

```python
# 6. Heatmap
# Heatmap is used to visualize matrix-like data using colors.
# It is commonly used for correlation matrices.

# Example: Correlation heatmap

corr=tips.corr(numeric_only=True)
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.show()

# Explanation: annot=True displays values and cmap controls color scheme
```

```python
# 7. Pairplot
# Pairplot is used to plot pairwise relationships between multiple numerical variables.

# Example: Pairwise relationships in dataset

iris=sns.load_dataset("iris")
sns.pairplot(iris,hue="species")
plt.show()

# Explanation: hue adds category distinction
```
